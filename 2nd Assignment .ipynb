```python
# 1. Student Grade System
# Marks < 35 → Fail
# Marks ≥ 35 and < 75 → Grade C
# Marks ≥ 75 and < 90 → Grade B
# Marks ≥ 90 → Grade A

marks = float(input("Enter the marks :"))
if(marks>=35):
    if(marks<90):
        if(marks<70):
            print("grade c")
        else:
            print("grade b")
    else:
        print("grade a")
else:
    print("fail")
```

```python
# 2. Number Check
# Number divisible by 2 → Even
# If even and divisible by 4 → Special Even
# Number not divisible by 2 → Odd
# If odd and divisible by 3 → Special Odd

num = int(input("Enter a Number: "))
if num % 2 == 0:
    if num % 4 == 0:
        print("Special Even")
    else:
        print("Even")
else:
    if num % 3 == 0:
        print("Special Odd")
    else:
        print("Odd")
```

```python
# 3. Age Classification
# Age < 18 → Minor
# Age ≥ 18 and < 60 → Adult
# Age ≥ 60 → Senior Citizen

age = int(input("Enter your age: "))
if(age>=18):
    if(age<60):
        print("Adult")
    else:
        print("Senior Citizen")
else:
    print("Minor")
```

```python
# 4. Temperature Category
# Temp < 15 → Cold
# Temp ≥ 15 and ≤ 30 → Moderate
# Temp > 30 and ≤ 40 → Hot
# Temp > 40 → Very Hot

temp = float(input("Enter the temperature: "))
if(temp<=30):
    if(temp>15):
        print("Cold")
    else:
        print("Moderate")
elif(temp<40):
    print("Hot")
else:
    print("Very hot")
```

```python
# 5. ATM Withdrawal
# Withdrawal amount ≤ balance → Proceed
# Withdrawal amount must be multiple of 100
# Otherwise → Transaction not allowed

balance = 5000 
withdrawal = int(input("Enter amount to withdraw: "))

if (withdrawal % 100 == 0):
    
    if (withdrawal <= balance):
        print("Proceed")
        balance -= withdrawal
        print(f"Remaining balance: {balance}")
    else:
        print("Transaction not allowed: Insufficient funds.")

else:
    print("Transaction not allowed: Amount must be a multiple of 100.")
```

```python
# 7. Electricity Bill
# Units ≤ 50 → ₹2/unit
# Units > 50 and ≤ 100 → ₹3/unit
# Units > 100 and ≤ 200 → ₹5/unit
# Units > 200 → ₹8/unit

unit = float(input("Enter the total units : "))
if(unit >=50):
    if(unit >= 100):
        if(unit <=200):
            print("₹5/unit")
        else:
            print("₹8/unit")
    else:
        print("₹3/unit")
else:
    print("₹2/unit")

```

```python
# 8. Shopping Discount
# Amount < 500 → No discount
# Amount ≥ 500 and < 1000 → 5% discount
# Amount ≥ 1000 and < 5000 → 10% discount
# Amount ≥ 5000 → 20% discount

amount = float(input("Enter the amount :"))
if(amount >= 500):
    if(amount >= 1000):
        if(amount >= 5000):
            print("20% discount - Total amount :" , amount-(amount * 0.2))
        else:
            print("10% discount - Total amount :", amount-(amount * 0.1))
    else:
        print("5% discount - Total amount :", amount-(amount * 0.05))
else:
    print("No discount - Total amount _", amount)
            
```

```python
# 9. Loan Eligibility
# Salary ≥ 25000 and credit score ≥ 750 → Approved
# Salary ≥ 25000 and credit score < 750 → Review
# Salary < 25000 and credit score ≥ 800 → Approved
# Otherwise → Rejected

sal = float(input("Enter the Amount :"))
cre_sco = float(input("Enter the Credit Score :"))
if(sal >= 25000 and cre_sco >= 750):
    if(sal >= 25000 and cre_sco > 750):
        if(sal >= 25000 and cre_sco >= 800):
            print("Approved")
        else:
            print("Review")
    else:
         print("Approved")
else:
    print("Rejected")



```

```python
# 10. Exam Result System
# All subjects ≥ 35 → Pass
# Any subject < 35 → Fail
# If pass:
# Average ≥ 85 → Distinction
# Average ≥ 60 and < 85 → First Class
# Average < 60 → Pass Class
# If any subject < 20 → Fail (Severe)

s1 = int(input("Subject 1: "))
s2 = int(input("Subject 2: "))
s3 = int(input("Subject 3: "))
s4 = int(input("Subject 4: "))
s5 = int(input("Subject 5: "))

if (s1 >= 35 and s2 >= 35 and s3 >= 35 and s4>=35 and s4>=35):
    
    avg = (s1 + s2 + s3+s4+s5) / 5
    print(f"Average: {avg:.2f}")
    if avg >= 60:
         if avg >= 85:
             print("Status: Distinction")
         else:
             print("Status: First Class")
    else:
        print("pass class")
else:
    if s1 < 20 or s2 < 20 or s3 < 20 or s4 < 20 or s5 < 20:
        print("Status: Fail (Severe)")
    else:
        print("Status: Fail")
```

```python
# 11. Ride Fare Calculation
# Distance ≤ 5 km and non-peak → ₹10/km
# Distance ≤ 5 km and peak → ₹15/km
# Distance > 5 km and non-peak → ₹8/km
# Distance > 5 km and peak → ₹12/km

distance = float(input("Enter distance in km: "))
peak = input("Is it peak hour? (yes/no): ")

if (distance <= 5):
    if peak == "yes":
        rate = 15
    else:
        rate = 10
else:
    
    if peak == "yes":
        rate = 12
    else:
        rate = 8

total_fare = distance * rate

print(f"Rate per km: {rate}")
print(f"Total Fare: {total_fare}")
```

```python
# 14. Movie Ticket Pricing
# Age < 12 → ₹100
# Age ≥ 12 and < 60 → ₹200
# Age ≥ 60 → ₹150
# Evening show → Additional ₹50

age = int(input("Enter your age: "))
show_time = input("Is it an evening show? (y/n): ")

if (age >= 60):
    if (show_time == "y"):
        price = 150 + 50
    else:
        price = 150
else:
    if (age >= 12):
        if show_time == "y":
            price = 200 + 50
        else:
            price = 200
    else:
        if (show_time == "y"):
            price = 100 + 50
        else:
            price = 100

print(f"Total Ticket Price: {price}")
```
